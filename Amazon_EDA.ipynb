# **Project Summary -**

Amazon Prime Video is one of the major global streaming platforms, offering a diverse collection of movies and television shows across different genres, countries, release years, age certifications, and content formats. The objective of this project is to perform an Exploratory Data Analysis (EDA) on the Amazon Prime catalog to identify meaningful patterns and trends in the available content. The analysis uses two datasets: title.csv, which contains information about movies and TV shows, and credit.csv, which contains information about actors, directors, and other people associated with the content.

The analysis begins with data loading, inspection, and cleaning. The datasets are examined for their dimensions, data types, duplicate records, missing values, and unique values. The visualization phase follows the UBM approach, consisting of Univariate, Bivariate, and Multivariate Analysis. Univariate analysis is used to understand the distribution of content types, release years, genres, ratings, runtime, IMDb scores, and seasons. Bivariate analysis examines relationships such as content type versus rating, genre versus content type, country versus content volume, release year versus content type, and runtime versus IMDb score. Multivariate analysis is used to understand more complex relationships among year, genre, content type, rating, runtime, and IMDb performance.

The analysis also explores the top countries producing Amazon Prime content, the most common genres, frequently credited actors and directors, and trends in content releases over time. A heatmap is used to identify the relationship between release years and genres, while scatter plots help examine relationships between runtime, popularity, votes, and ratings.

The insights generated from this analysis can help streaming platforms understand content preferences, identify popular genres, evaluate regional content availability, and observe how the content catalog has evolved over time. The analysis can also support decisions related to content acquisition, production, regional expansion, and audience targeting.

Overall, this project provides a comprehensive overview of the Amazon Prime catalog and demonstrates how exploratory data analysis and visualization can transform raw entertainment data into meaningful business insights.

## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Write your code to make your dataset analysis ready.
df = titles_df.copy()
credits = credits_df.copy()
```

```python
df.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)

print("Title dataset shape after removing duplicates:", df.shape)
print("Credit dataset shape after removing duplicates:", credits.shape)
```

```python
credits['release_year'] = pd.to_numeric(credits['release_year'], errors='coerce')
```

```python
numeric_columns = [
    'runtime',
    'seasons',
    'imdb_score',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score'
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(df[column], errors='coerce')
```

```python
text_columns = ['title', 'type', 'age_certification']

for column in text_columns:
    if column in credits.columns:
        credits[column] = credits[column].astype('string').str.strip()

credits['type'] = credits['type'].str.upper()
```

```python
def parse_list_column(value):
    if pd.isna(value):
        return []

    try:
        result = ast.literal_eval(value)
        if isinstance(result, list):
            return [str(x).strip().lower() for x in result]
        return []
    except:
        return []

credits['genres_list'] = credits['genres'].apply(parse_list_column)
```

```python
credits['countries_list'] = credits['production_countries'].apply(parse_list_column)
```

```python
credits['age_certification'] = credits['age_certification'].fillna('Unknown')

credits['runtime'] = credits['runtime'].fillna(credits['runtime'].median())

credits['imdb_score'] = credits['imdb_score'].fillna(credits['imdb_score'].median())

credits['tmdb_score'] = credits['tmdb_score'].fillna(credits['tmdb_score'].median())

credits['tmdb_popularity'] = credits['tmdb_popularity'].fillna(
    credits['tmdb_popularity'].median()
)

credits['imdb_votes'] = credits['imdb_votes'].fillna(
    credits['imdb_votes'].median()
)
```

```python
credits['seasons'] = credits['seasons'].fillna(0)
```

```python
print("Missing values after cleaning:")
display(df.isnull().sum().sort_values(ascending=False))
```

### What all manipulations have you done and insights you found?

The dataset was cleaned by removing duplicate records, standardizing categorical text values, converting numerical columns into appropriate numeric data types, and converting list-like genre and country columns into Python lists. Missing values were handled using appropriate strategies such as median imputation for numerical variables and "Unknown" for missing age certifications. The credit dataset was also checked for duplicates.

The cleaned data is now suitable for statistical analysis and visualization. Separating genres and countries into lists allows individual genres and countries to be analyzed accurately rather than treating the entire list as one category.
